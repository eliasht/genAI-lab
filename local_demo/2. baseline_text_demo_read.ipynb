# Query with metadata filter

```python
from llama_index.core.vector_stores import (
    MetadataFilter,
    MetadataFilters,
    FilterOperator,
)
from llama_index.core.postprocessor import SimilarityPostprocessor
from llama_index.core.query_engine import RetrieverQueryEngine

filters = MetadataFilters(
    filters=[
        MetadataFilter(
            key="dokumen", operator=FilterOperator.EQ, value="pajak"
        ),
    ]
)

policy_query_engine = index...(filters = filters)
```

```python
response = policy_query_engine.query(
    "Penghasilan apa saja yang dikecualikan dari objek pajak?",
)
```

```python
# Check response text

print(response.response)
```

```python
# Check retrieved nodes
response.source_nodes
```

```python
nodes = response.source_nodes
for i, node_with_score in enumerate(nodes, start=1):
    node = node_with_score.node
    metadata = node.metadata
    text = node.text.strip()
    score = node_with_score.score

    print(f"=== Node {i} ===")
    print("Score   :", round(score, 4))
    print("File    :", metadata.get("file_name"))
    print("Halaman :", metadata.get("page_label"))
    print("Dokumen :", metadata.get("dokumen"))
    print("-" * 40)
    print(text)  
    print("\n")
```
